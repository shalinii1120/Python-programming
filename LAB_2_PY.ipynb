 1.Prime Number or Not a Prime Number.


```python

# Program to check whether a number is Prime or Not

num = int(input("Enter a positive integer: "))

if num <= 1:
    print(num, "is Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is Not a Prime Number")
            break
    else:
        print(num, "is a Prime Number")
```

2.the factorial of a positive integer using loop.


```python

# Program to find factorial using a loop

num = int(input("Enter a positive integer: "))

factorial = 1

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    for i in range(1, num + 1):
        factorial = factorial * i

    print("Factorial of", num, "is:", factorial)
```

3.OTP GENERATOR



```python

# OTP Generator

import random

n = int(input("Enter the number of digits required for OTP (4 or 6): "))

if n == 4 or n == 6:
    otp = random.randint(10**(n-1), 10**n - 1)
    print("Generated OTP:", otp)
else:
    print("Please enter only 4 or 6.")
```

5.Fibonacci series using recursion

```python

# Fibonacci Series using Recursion

def fibonacci(n):
    if n == 0 or n == 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


n = int(input("Enter number of terms: "))

for i in range(n):
    print(fibonacci(i), end=" ")
```
