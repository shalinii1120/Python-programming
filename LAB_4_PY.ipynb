1.Log file Reader

```python

# Log File Reader and Error Counter

try:
    with open("log.txt", "r") as file:
        error_count = 0
        line = file.readline()

        while line:
            print(line.strip())

            if "Error" in line:
                error_count += 1

            line = file.readline()

        print(f"Total errors found: {error_count}")

except FileNotFoundError:
    print("The file 'log.txt' was not found.")
```

2.Data Manager Using Files

```python

# Student Data Manager using File Handling

while True:
    print("\n----- STUDENT DATA MANAGER -----")
    print("1. Add a new student record")
    print("2. Display all student records")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        roll_no = input("Enter Roll Number: ")
        name = input("Enter Student Name: ")
        marks = input("Enter Marks: ")

        with open("students.txt", "a") as file:
            file.write(roll_no + "," + name + "," + marks + "\n")

        print("Student record added successfully!")

    elif choice == "2":
        try:
            with open("students.txt", "r") as file:
                print("\n----- STUDENT RECORDS -----")
                records = file.readlines()

                if len(records) == 0:
                    print("No student records found.")
                else:
                    for record in records:
                        roll_no, name, marks = record.strip().split(",")

                        print("Roll Number:", roll_no)
                        print("Student Name:", name)
                        print("Marks:", marks)
                        print("--------------------------")

        except FileNotFoundError:
            print("No student records found. File does not exist.")

    elif choice == "3":
        print("Exiting the program...")
        break

    else:
        print("Invalid choice! Please try again.")
```

3.based payroll management system

```python

# Employee Payroll Management System

import csv

try:
    n = int(input("Enter number of employees: "))

    with open("payroll.csv", "w", newline="") as file:
        writer = csv.writer(file)

        writer.writerow(["ID", "Name", "Basic", "Allowance",
                         "Deduction", "Net Salary"])

        for i in range(n):
            print("\nEnter Employee Details")

            emp_id = input("Employee ID: ")
            name = input("Employee Name: ")
            basic = float(input("Basic Salary: "))
            allowance = float(input("Allowance: "))
            deduction = float(input("Deduction: "))

            net_salary = basic + allowance - deduction

            writer.writerow([emp_id, name, basic, allowance,
                             deduction, net_salary])

    print("\nPayroll details saved successfully in payroll.csv")

except ValueError:
    print("Invalid input. Please enter numeric values correctly.")

except Exception as e:
    print(f"Error: {e}")
```

4.Log
analyzer

```python
import json

# 1. Programmatically write the logs.json file
logs_data = [
    {"timestamp": "10:00", "level": "INFO", "message": "Application started"},
    {"timestamp": "10:05", "level": "WARNING", "message": "Low memory detected"},
    {"timestamp": "10:10", "level": "ERROR", "message": "Database connection failed"},
    {"timestamp": "10:15", "level": "INFO", "message": "User logged in"},
    {"timestamp": "10:20", "level": "ERROR", "message": "File not found"},
    {"timestamp": "10:25", "level": "WARNING", "message": "High CPU usage"}
]

with open("logs.json", "w") as file:
    json.dump(logs_data, file, indent=4)

# 2. Log Analyzer main.py logic
try:
    with open("logs.json", "r") as file:
        data = json.load(file)

    if not isinstance(data, list):
        raise ValueError("Invalid JSON data: Expected a list of logs.")

    info_count = 0
    warning_count = 0
    error_count = 0

    print("\n----- ALL LOG RECORDS -----")
    for log in data:
        if not isinstance(log, dict):
            print("Invalid log record:", log)
            continue

        timestamp = log.get("timestamp", "N/A")
        level = log.get("level", "N/A")
        message = log.get("message", "N/A")

        print("Timestamp:", timestamp)
        print("Level:", level)
        print("Message:", message)
        print("--------------------------")

        level = str(level).upper()
        if level == "INFO":
            info_count += 1
        elif level == "WARNING":
            warning_count += 1
        elif level == "ERROR":
            error_count += 1

    print("\n----- LOG SUMMARY -----")
    print("Total INFO messages:", info_count)
    print("Total WARNING messages:", warning_count)
    print("Total ERROR messages:", error_count)
    print("Total log records:", len(data))

except FileNotFoundError:
    print("Error: logs.json file not found.")
except json.JSONDecodeError:
    print("Error: Invalid JSON data in logs.json.")
except ValueError as e:
    print(e)
```
